```python
def calculate_bill(products):
    bill = 0
    for name in products:
        price = products[name]["price"]
        quantity = products[name]["quantity"]
        total_price = price * quantity
        products[name]["total"] = total_price
        bill = bill + total_price
    return bill


products = {}

for i in range(1, 6):
    product_name = input("Enter the name of product : ")
    product_price = float(input("Enter the price of product : "))
    product_quantity = int(input("Enter the quantity of product : "))

    products[product_name] = {
        "price": product_price,
        "quantity": product_quantity,
        "total": 0
    }

bill = calculate_bill(products)

# Apply 10% discount if the bill is greater than 10000
discount = 0
if bill > 10000:
    discount = bill * 0.10

final_bill = bill - discount

# Find the most expensive product
expensive_name = ""
expensive_price = 0
for name in products:
    if products[name]["price"] > expensive_price:
        expensive_price = products[name]["price"]
        expensive_name = name

# Display on screen
print("\n------ BILL ------")
for name in products:
    print("Product Name:", name)
    print("Price:", products[name]["price"])
    print("Quantity:", products[name]["quantity"])
    print("Total Price:", products[name]["total"])
    print()

print("Complete Bill:", bill)
print("Discount:", discount)
print("Final Bill:", final_bill)
print("Most Expensive Product:", expensive_name, "- Rs.", expensive_price)

# Save the bill in bill.txt
file = open("bill.txt", "w")
file.write("------ BILL ------\n")
for name in products:
    file.write("Product Name: " + name + "\n")
    file.write("Price: " + str(products[name]["price"]) + "\n")
    file.write("Quantity: " + str(products[name]["quantity"]) + "\n")
    file.write("Total Price: " + str(products[name]["total"]) + "\n\n")
file.write("Complete Bill: " + str(bill) + "\n")
file.write("Discount: " + str(discount) + "\n")
file.write("Final Bill: " + str(final_bill) + "\n")
file.write("Most Expensive Product: " + expensive_name + " - Rs. " + str(expensive_price) + "\n")
file.close()

print("Bill has been saved in bill.txt.")
```
