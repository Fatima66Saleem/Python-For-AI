```python
subjects = ["python", "Artificial Intelligence", "Mathematics"]

students = []

for i in range(1, 5):
    print(f"\n--- Student {i} ---")
    name = input("Enter the student name: ")
    roll_no = input("Enter student roll no: ")

    marks = {}
    for sub in subjects:
        sub_marks = int(input(f"Enter marks for {sub}: "))
        marks[sub] = sub_marks

    student_info = {
        "name": name,
        "roll_no": roll_no,
        "marks": marks
    }

    students.append(student_info)


def marks_calculation(student, total_per_subject=100):
    total_marks = 0
    for mark in student["marks"].values():
        total_marks += mark

    max_marks = total_per_subject * len(student["marks"])
    percentage = (total_marks / max_marks) * 100

    if percentage >= 80 and percentage <= 100:
        grade = "A"
    elif percentage >= 70 and percentage < 80:
        grade = "B"
    elif percentage >= 60 and percentage < 70:
        grade = "C"
    elif percentage >= 50 and percentage < 60:
        grade = "D"
    else:
        grade = "F"

    return total_marks, percentage, grade


# result.txt kholna ("w" = write mode)
with open("result.txt", "w") as file:

    # Har student ka result print karna, dictionary ma save karna aur file ma likhna
    for student in students:
        result_total_marks, result_percentage, result_grade = marks_calculation(student)

        student["total"] = result_total_marks
        student["percentage"] = result_percentage
        student["grade"] = result_grade

        print("\n")
        print("Name: ", student["name"])
        print("Roll no: ", student["roll_no"])
        for sub, mark in student["marks"].items():
            print(f"{sub}: {mark}")
        print("Total Marks: ", result_total_marks)
        print(f"Percentage: {result_percentage:.2f}%")
        print("Grade: ", result_grade)

        # Wohi result file ma likhna
        file.write(f"Name: {student['name']}\n")
        file.write(f"Roll no: {student['roll_no']}\n")
        for sub, mark in student["marks"].items():
            file.write(f"{sub}: {mark}\n")
        file.write(f"Total Marks: {result_total_marks}\n")
        file.write(f"Percentage: {result_percentage:.2f}%\n")
        file.write(f"Grade: {result_grade}\n")
        file.write("-" * 30 + "\n")

    # ---------- Highest marks wala student ----------
    topper = students[0]

    for student in students:
        if student["total"] > topper["total"]:
            topper = student

    print("\n===== Topper =====")
    print("Name: ", topper["name"])
    print("Roll no: ", topper["roll_no"])
    print("Total Marks: ", topper["total"])
    print(f"Percentage: {topper['percentage']:.2f}%")
    print("Grade: ", topper["grade"])

    # Topper ko bhi file ma likhna
    file.write("\n===== Topper =====\n")
    file.write(f"Name: {topper['name']}\n")
    file.write(f"Roll no: {topper['roll_no']}\n")
    file.write(f"Total Marks: {topper['total']}\n")
    file.write(f"Percentage: {topper['percentage']:.2f}%\n")
    file.write(f"Grade: {topper['grade']}\n")

print("\nResults result.txt ma save ho gaye hain.")
```
