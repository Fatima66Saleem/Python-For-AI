```python
sentence = input("Enter a sentence: ")
words = sentence.split()
print("The number of words in the sentence is:", len(words))
print("The number of characters in the sentence is:", len(sentence))
longest_word = max(words, key=len)
print("The longest word in the sentence is:", longest_word)
shortest_word = min(words, key=len)
print("The shortest word in the sentence is:", shortest_word)
vowels = "aeiouAEIOU"
count = 0
for char in sentence:
    if char in vowels:
        count += 1
print("The number of vowels in the sentence is:", count)

unique_words = set(words)
print("The number of unique words in the sentence is:", len(unique_words))

word_alphabetical = sorted(words)
print("The words in alphabetical order are:", word_alphabetical)

with open("word_analysis.txt", "w") as f:
    f.write("The number of words in the sentence is: " + str(len(words)) + "\n")
    f.write("The number of characters in the sentence is: " + str(len(sentence)) + "\n")
    f.write("The longest word in the sentence is: " + longest_word + "\n")
    f.write("The shortest word in the sentence is: " + shortest_word + "\n")
    f.write("The number of vowels in the sentence is: " + str(count) + "\n")
    f.write("The number of unique words in the sentence is: " + str(len(unique_words)) + "\n")
    f.write("The words in alphabetical order are: " + str(word_alphabetical) + "\n")
```
