```python
i = 0
count = 0
num = []

while i < 5:
    number = int(input("Enter a number: "))
    num.append(number)
    i += 1

for item in num:
    if item % 2 == 0:
        print(item, "is even")
        count += 1
    else:
        print(item, "is odd")

print("Total even numbers:", count)
        
```
