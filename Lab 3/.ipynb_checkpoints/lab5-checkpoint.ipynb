```python
def deposit(amount):
    total_amount = 0
    total_amount += amount
    return total_amount
    
def withdraw(total_balance, withdraw_amount):
    if total_balance > 0:
        if total_balance >= withdraw_amount:
            total_balance -= withdraw_amount
            return total_balance
        else:
            print("Insufficiant balance")
    else:
        print("Invalid entry")
    
def check_balance(result_withdraw):
    return result_withdraw

def account(result_deposit, withdraw_amount, result_withdraw):
    print("Amount deposit:",result_deposit)
    print("Amount Withdraw:", withdraw_amount)
    print("Total Balance:", result_withdraw)
    
deposit_amount = int(input("Enter the deposit amount: "))
if deposit_amount > 0:
    result_deposit = deposit(deposit_amount)
    print(deposit_amount,"value deposited: ",result_deposit)
else:
    print("Invalid ammount")
    
withdraw_amount = int(input("Enter the withdraw amount: "))
if result_deposit > 0:
    if withdraw_amount <= result_deposit:
        result_withdraw = withdraw(result_deposit, withdraw_amount)
        print(result_withdraw)
    else:
        print("insufficiant balance")
else:
    print("Invalid")

balance = check_balance(result_withdraw)
print("Total balance is: ",balance)

account(result_deposit, withdraw_amount, result_withdraw)
```
