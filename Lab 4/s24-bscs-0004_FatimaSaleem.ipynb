```python
# task 2
num = int(input("Enter a number: "))
reversed_number = 0
while num > 0:
    digit = num % 10
    reversed_number = reversed_number * 10 + digit
    num = num // 10
print(reversed_number)
```

```python
# task 3
i = 0
count = 0
num = [] # intialize empty list

while i < 5:
    number = int(input("Enter a number: "))
    num.append(number)
    i += 1
for item in num:
    if item % 2 == 0:
        print(item, "is even")
        count += 1
    else:
        print(item, "is odd")

print("Total even numbers:", count)
```

```python
marks = int(input("Enter your marks: "))

if marks > 100 or marks < 0:
    print("Invalid marks!")
elif marks >= 90:
    print("Grade: A+")
elif 80 <= marks < 90:
    print("Grade: A")
elif 75 <= marks < 80:
    print("Grade: B+")
elif 70 <= marks < 75:
    print("Grade: B")
elif 65 <= marks < 70:
    print("Grade: C+")
elif 60 <= marks < 65:
    print("Grade: C")
elif 55 <= marks < 60:
    print("Grade: D+")
elif 50 <= marks < 55:
    print("Grade: D")
else:
    print("Grade: F")
```

```python
student_info = {
    "name": "Fatima",
    "age": 22,
    "course": "Theory of Automata",
    "marks": 91
}
print("Original dictionary:", student_info)


print("Name:", student_info["name"])
print("Marks:", student_info["marks"])

student_info["marks"] = 95 # update only sinle value at time
student_info["age"] = 23

print("Updated dictionary:", student_info)

student_info.update({ # update multiple values 
    "name": "Aqsa", 
    "age": 21
})
print("Updated dictionary:", student_info)
```

```python
student_info = {
    "name": "Fatima",
    "age": 22,
    "course": "Theory of Automata",
    "marks": 91
}
print("keys():", student_info.keys())

print("values():", student_info.values())

print("items():", student_info.items())

print("get(course):", student_info.get("course"))
print("get(grade, A):", student_info.get("grade", "A"))

student_info.update({"marks": 95, "grade": "A"})
print("after update():", student_info)

removed = student_info.pop("age")
print("popped value:", removed)
print("after pop():", student_info)
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# union()
print("union():", set1.union(set2))

# intersection()
print("intersection():", set1.intersection(set2))

# difference()
print("difference():", set1.difference(set2))

# symmetric_difference()
print("symmetric_difference():", set1.symmetric_difference(set2))
```
